# Dashboard Data Generation

```python
import pandas as pd

node_metrics = pd.read_csv(
    r"D:\Projects\Delivery_ETA\outputs\node_metrics.csv"
)

top_bottlenecks = pd.read_csv(
    r"D:\Projects\Delivery_ETA\outputs\top_bottlenecks.csv"
)

corridor_metrics = pd.read_csv(
    r"D:\Projects\Delivery_ETA\outputs\corridor_metrics.csv"
)

comparison = pd.read_csv(
    r"D:\Projects\Delivery_ETA\outputs\model_comparison.csv"
)

feature_importance = pd.read_csv(
    r"D:\Projects\Delivery_ETA\outputs\feature_importance.csv"
)

print("Loaded Successfully")
```

## Dashboard Hub Data

```python
dashboard_hubs = (
    top_bottlenecks[
        [
            "source_center",
            "trips",
            "breach_rate",
            "betweenness_centrality"
        ]
    ]
    .copy()
)

dashboard_hubs.head()
```

## Dashboard Corridor Data

```python
dashboard_corridors = (
    corridor_metrics
    .sort_values(
        "delay_ratio",
        ascending=False
    )
    .head(50)
)

dashboard_corridors.head()
```

## Dashboard Model Metrics

```python
dashboard_metrics = comparison.copy()

dashboard_metrics
```

```python
print(dashboard_hubs.shape)
print(dashboard_corridors.shape)
print(dashboard_metrics.shape)
```

## Dashboard Feature Importance

```python
dashboard_features = (
    feature_importance
    .head(15)
)

dashboard_features.head()
```

```python
dashboard_hubs.to_csv(
    r"D:\Projects\Delivery_ETA\outputs\dashboard_hubs.csv",
    index=False
)

dashboard_corridors.to_csv(
    r"D:\Projects\Delivery_ETA\outputs\dashboard_corridors.csv",
    index=False
)

dashboard_metrics.to_csv(
    r"D:\Projects\Delivery_ETA\outputs\dashboard_model_metrics.csv",
    index=False
)

dashboard_features.to_csv(
    r"D:\Projects\Delivery_ETA\outputs\dashboard_feature_importance.csv",
    index=False
)

print("Saved Successfully")
```
