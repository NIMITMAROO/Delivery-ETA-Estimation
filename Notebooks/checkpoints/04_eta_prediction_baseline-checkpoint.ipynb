# Baseline ETA Prediction Model

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor

leg_df = pd.read_csv(
    r"D:\Projects\Delivery_ETA\outputs\leg_level_data.csv"
)

print(leg_df.shape)
```

## Feature Preparation

```python
model_df = leg_df.copy()

model_df["route_type"] = (
    model_df["route_type"]
    .map({
        "FTL": 1,
        "Carting": 0
    })
)

features = [
    "osrm_time",
    "osrm_distance",
    "actual_distance_to_destination",
    "route_type"
]

X = model_df[features]

y = model_df["actual_time"]

print(X.shape)
print(y.shape)
```

## Train-Test Split

```python
X_train, X_test, y_train, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )
)

print(X_train.shape)
print(X_test.shape)
```

## Baseline Model Training

```python
baseline_model = XGBRegressor(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.05,
    random_state=42
)

baseline_model.fit(
    X_train,
    y_train
)

print("Training Complete")
```

## Baseline Predictions

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

y_pred = baseline_model.predict(X_test)

print("Predictions Generated")
```

```python
mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2 = r2_score(
    y_test,
    y_pred
)

print("MAE :", round(mae,2))
print("RMSE:", round(rmse,2))
print("R2  :", round(r2,4))
```
