# Model Comparison

```python
import pandas as pd

comparison = pd.DataFrame({
    "Model": [
        "Baseline XGBoost",
        "Graph Enhanced XGBoost"
    ],
    "MAE": [
        47.12,
        31.49
    ],
    "RMSE": [
        108.40,
        90.92
    ],
    "R2": [
        0.9297,
        0.9506
    ],
    "Accuracy_15pct": [
        0,
        54.46
    ]
})

comparison
```

```python
comparison["MAE_Improvement_%"] = (
    (
        47.12 - comparison["MAE"]
    )
    / 47.12
    * 100
)

comparison
```

```python
comparison.to_csv(
    r"D:\Projects\Delivery_ETA\outputs\model_comparison.csv",
    index=False
)

comparison
```

```python
print(
    "MAE Improvement:",
    round(
        ((47.12 - 31.49) / 47.12) * 100,
        2
    ),
    "%"
)
```
